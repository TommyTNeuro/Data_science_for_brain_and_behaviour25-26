# Supplementary Guide : Linux

This guide is for those interested in learning about some additional programming concepts and tools that are adjacent to but not necessary to know to pass the module. These guides will cover background information, their place in the programming world, practical advice on how you can use these practically in your programming career. This is a part of a set of guides - this guide is particularly on linux.

The main way you'll interact with linux in the module is through vlab mainly through the terminal. Vlab is running on a "linux virtual machine", which essentially means a computer running somewhere else is running linux and you are accessing it through the web. Pretty neat right?
